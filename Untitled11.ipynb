```python
def increment_string(strng):
    head = strng.rstrip('0123456789')
    tail = strng[len(head):]
    if tail == "": return strng+"1"
    return head + str(int(tail) + 1).zfill(len(tail))
```

```python
def increment_string(string):
    # Start checking from the end of the string
    i = len(string)

    # Move backwards as long as the character is a digit
    while i > 0 and string[i-1].isdigit():
        i -= 1

    # If the index didn't move, there are no numbers at the end
    if i == len(string):
        return string + '1'

    # Split the string into the text part and the number part
    text_part = string[:i]
    number_str = string[i:]

    # Increment the number and pad with leading zeros using zfill
    new_number = str(int(number_str) + 1).zfill(len(number_str))

    return text_part + new_number
```
